Problem 1 Buffalo Wings

image.png

```python
# Loading the Gurobipy library
import gurobipy as gp

# Create the model
model = gp.Model("Wings")

# Create the decision variables
x1 = model.addVar(vtype=gp.GRB.CONTINUOUS)
x2 = model.addVar(vtype=gp.GRB.CONTINUOUS)
x3 = model.addVar(vtype=gp.GRB.CONTINUOUS)
x4 = model.addVar(vtype=gp.GRB.CONTINUOUS)
x5 = model.addVar(vtype=gp.GRB.CONTINUOUS)
x6 = model.addVar(vtype=gp.GRB.CONTINUOUS)

# Create the constraints
model.addConstr(20*x1 + 20*x2 + 18*x3 + 22*x4 + 20*x5 <= 4000)  # Bouillon mix, lbs
model.addConstr(0.2*x1 + 0.3*x2 + 0.25*x3 + 0.3*x4 + 0.4*x5 <= 450) # sauce base, gals
model.addConstr(1.0*x1 <= 1200) # classic spice, lbs
model.addConstr(1.5*x2 <= 800) # honey bbq, lbs
model.addConstr(1.3*x3 <= 1000) # Garlic parm lbs
model.addConstr(1.7*x4 <= 400) # Spicy mango lbs
model.addConstr(1.0*x1 + 1.5*x2 + 1.2*x3 + 1.5*x4 + 2.0*x5 <= 300) # labour

# By default the variables are non negative

# Create Objective function
model.setObjective(60*x1 + 75*x2 + 70*x3 + 90*x4 + 110*x5, gp.GRB.MAXIMIZE)

# Optimize the model
model.optimize()

# Print the variables
print(f"The values of x1 is {x1.x}")
print(f"The values of x2 is {x2.x}")
print(f"The values of x3 is {x3.x}")
print(f"The values of x4 is {x4.x}")
print(f"The values of x5 is {x5.x}")
print(f"The values of x6 is {x6.x}")




```

Report:
Max Profit can be earn by making ghost peppers and spicy mango sauce and not making the the others. In a real restaurant there maybe a need to have a minimum amount of the other sauces as wel lbut this was not mentioned in the question. Also there is no constraint on the amount of ghost peppers spice mix available so  it is assumed to be unlimited for this problem. Although it looks like making maximum ghost peppers would be most profitable (110 per gal) it uses more labour hours and hence the balance is 142.86 gal of spicy mango and 42.86 gals of ghost pepper sauce for a max profit of 12571.43 USD.

Problem 2 Buffalo Bisons

image.png

```python
# Create the model
model = gp.Model("Bisons")

# Create the decision variables
p1 = model.addVar(vtype=gp.GRB.CONTINUOUS)
p2 = model.addVar(vtype=gp.GRB.CONTINUOUS)
p3 = model.addVar(vtype=gp.GRB.CONTINUOUS)
p4 = model.addVar(vtype=gp.GRB.CONTINUOUS)
p5 = model.addVar(vtype=gp.GRB.CONTINUOUS)

# Create the constraints
model.addConstr(1200 - 40*p1 <= 800)
model.addConstr(2000 - 50*p2 <= 1500)
model.addConstr(3500 - 70*p3 <= 2600)
model.addConstr(4000 - 90*p4 <= 3000)
model.addConstr(300 - 15*p5 <= 250)
model.addConstr(p1 >= 15)
model.addConstr(p2 >= 12)
model.addConstr(p3 >= 8)
model.addConstr(p4 >= 5)
model.addConstr(p5 >= 25)

# By default the variables are non negative on Gurobipy

# Create Objective function
model.setObjective((1200*p1 - 40*p1*p1) + (2000*p2 - 50*p2*p2) + (3500*p3 - 70*p3*p3) + (4000*p4 - 90*p4*p4) + (300*p5 - 15*p5*p5), gp.GRB.MAXIMIZE)

# Optimize the model
model.optimize()

# Print the variables
print(f"The values of x1 is {p1.x}")
print(f"The values of x2 is {p2.x}")
print(f"The values of x3 is {p3.x}")
print(f"The values of x4 is {p4.x}")
print(f"The values of x5 is {p5.x}")
```

Report:
Report: According to this Buffalo Bisons can make maximum revenue from ticket sales by selling field at 15, terrace at 20, reserved at 25, bleachers at 22.22 and the party box at the minimum required by management at 25 giving 115319.44 USD in revenue. If the minimum ticket prices for the party deck were more flexible we may be able to get a higher revenue or as it stands we can keep the price at 25 and hope customers would be interested in purchasing it anyway.

Problem 3 Concourse

image.png

```python
# Create the model
model = gp.Model("Concourse")

# create decision variables
x1 = model.addVar(vtype = gp.GRB.BINARY)
x2 = model.addVar(vtype = gp.GRB.BINARY)
x3 = model.addVar(vtype = gp.GRB.BINARY)
x4 = model.addVar(vtype = gp.GRB.BINARY)
x5 = model.addVar(vtype = gp.GRB.BINARY)
x6 = model.addVar(vtype = gp.GRB.BINARY)

# Create the constraints
model.addConstr(80*x1 + 120*x2 + 95*x3 + 150*x4 + 60*x5 + 110*x6 <= 400)

# by default the variables are non neg on gurobi

# Create the objective function
model.setObjective(70*x1 + 85*x2 + 78*x3 + 95*x4 + 55*x5 + 80*x6 + 15*x1*x4 + 10*x2*x6 - 12*x3*x5, gp.GRB.MAXIMIZE)

# Optimize the model
model.optimize()

# Print the variables
print(f" Value of x1: {x1.x}")
print(f" Value of x2: {x2.x}")
print(f" Value of x3: {x3.x}")
print(f" Value of x4: {x4.x}")
print(f" Value of x5: {x5.x}")
print(f" Value of x6: {x6.x}")
```

Report:
The total value score can be maximised by funding projects P1 P4 P5 and P6 for a max value score of 315. Whole of the 400k USD will be used.
